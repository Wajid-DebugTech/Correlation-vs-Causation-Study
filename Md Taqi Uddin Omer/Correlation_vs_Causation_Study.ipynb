
# Correlation vs Causation Study

## Objective
This study demonstrates the difference between correlation and causation using the Diabetes Dataset from Scikit-Learn.

The goal is to:
- Identify relationships between variables
- Visualize correlations
- Understand why correlation does not always imply causation
- Interpret business implications from analytical findings



# Step 1: Dataset Selection and Understanding

## Why this dataset?
The Diabetes dataset contains multiple medical variables and a target variable called `disease_progression`.

### Features:
- age → Age of patient
- sex → Gender-related feature
- bmi → Body Mass Index
- bp → Blood pressure
- s1-s6 → Blood serum measurements

### Target:
- disease_progression → Quantitative measure of diabetes disease progression after one year

## Why this dataset is suitable?
This dataset contains multiple independent variables interacting with one dependent variable, making it ideal for:
- Correlation analysis
- Regression-style interpretation
- Understanding statistical relationships


```

from sklearn.datasets import load_diabetes
import pandas as pd

diabetes = load_diabetes(as_frame=True)
df = diabetes.frame

df.rename(columns={"target": "disease_progression"}, inplace=True)

df.head()
```


# Step 2: Correlation Analysis

## What is Correlation?
Correlation measures the strength and direction of a relationship between two variables.

### Correlation Coefficient Range
- +1 → Perfect positive correlation
- 0 → No correlation
- -1 → Perfect negative correlation

## Important Understanding
Correlation only indicates association.
It DOES NOT prove that one variable causes another.

Example:
- Ice cream sales and drowning incidents may both increase during summer.
- Ice cream does not cause drowning.
- Temperature is the hidden influencing variable.

## Why Correlation Heatmaps?
Heatmaps help visualize:
- Strong positive relationships
- Strong negative relationships
- Weak/no relationships


```

import matplotlib.pyplot as plt

corr = df.corr()

plt.figure(figsize=(10,8))
im = plt.imshow(corr, aspect='auto')
plt.colorbar(im)
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Matrix Heatmap")
plt.show()
```


# Step 3: Scatter Plot Analysis with Trend Lines

## Why Scatter Plots?
Scatter plots help visually inspect:
- Direction of relationship
- Strength of relationship
- Presence of outliers
- Potential linear trends

Trend lines summarize the overall direction.

## Variables Selected
### 1. BMI vs Disease Progression
Expected positive relationship.

### 2. Blood Pressure vs Disease Progression
Moderate relationship expected.

### 3. Serum Measurement (s5) vs Disease Progression
Typically strong relationship in diabetes analysis.


```

import numpy as np
import matplotlib.pyplot as plt

pairs = [
    ("bmi", "disease_progression"),
    ("bp", "disease_progression"),
    ("s5", "disease_progression")
]

for x, y in pairs:
    plt.figure(figsize=(6,4))
    plt.scatter(df[x], df[y])

    z = np.polyfit(df[x], df[y], 1)
    p = np.poly1d(z)

    plt.plot(df[x], p(df[x]))
    plt.xlabel(x)
    plt.ylabel(y)
    plt.title(f"{x} vs {y}")
    plt.show()
```


# Step 4: Correlation vs Causation

## Key Difference

### Correlation
Two variables move together statistically.

### Causation
One variable directly influences another.

## Why Correlation Alone is Dangerous
Several hidden factors can create misleading relationships:

### 1. Confounding Variables
A third variable influences both variables.

### 2. Reverse Causality
The outcome may actually influence the predictor.

### 3. Coincidental Relationships
Some relationships occur purely by chance.

## Example from This Study
BMI shows strong correlation with disease progression.

However:
- BMI alone may not directly cause progression.
- Lifestyle
- Genetics
- Diet
- Exercise
- Medication adherence

may also influence the outcome.

Therefore, correlation should be treated as:
- A signal for further investigation
- Not final proof of causality
